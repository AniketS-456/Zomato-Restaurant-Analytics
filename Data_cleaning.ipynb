```python
import pandas as pd
import numpy as np
import os

df = pd.read_csv('zomato.csv')
print (df.head)
print(df.head())
```

```python
columns_remove = ['url', 'address', 'phone', 'dish_liked', 'reviews_list', 'menu_item']
df = df.drop(columns=columns_remove, errors = 'ignore')
print(df.info())
```

```python
def clean_rate(value):
    if pd.isna(value) or value == 'NEW' or value == '-':
        return None
    number_part = str(value).split('/')[0].strip()
    return float(number_part)

df['rate'] = df['rate'].apply(clean_rate)
print(df.info())
df['rate'].unique()
```

```python
df['approx_cost(for two people)'].unique()
```

```python
def clean_cost(value):
    if pd.isna(value):
        return None
    cleaned_value = str(value).replace(',', '')
    return float(cleaned_value)

df['approx_cost(for two people)'] = df['approx_cost(for two people)'].apply(clean_cost)
print(df.info())
df['approx_cost(for two people)'].unique()
```

```python
df.duplicated().sum()
df = df.drop_duplicates()

print(df.info())
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=['location', 'rest_type', 'cuisines', 'approx_cost(for two people)'])
print(df.info())
print(df.shape)
```

```python
print(df['online_order'].unique())
print(df['book_table'].unique())
print(df['listed_in(type)'].unique())
```

```python
print(df['location'].nunique())
print(df['listed_in(city)'].nunique())
```

```python
df.to_csv("zomato_cleaned.csv", index=False)
```
